```python
# Step 1: Necessary libraries-ah import pannikalam
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob

# Step 2: Sample Social Media Dataset create pannuvom
data = {
    'Social_Media_Post': [
        "I absolutely love learning deep learning, it is amazing!",
        "This project execution is so bad and frustrating.",
        "Today is just a normal sunny day at the campus.",
        "Python is the best programming language for AI.",
        "The server is down again, I hate this slow response.",
        "Attended a great tech hackathon today, learned a lot!",
        "I am attending my regular engineering classes now.",
        "Terrible customer service, extremely disappointed."
    ]
}

df = pd.DataFrame(data)
print("--- Dataset Loaded Successfully ---")
print(df.head())

# Step 3: Sentiment Analysis functions (TextBlob using polarity)
def get_sentiment_polarity(text):
    return TextBlob(text).sentiment.polarity

def get_sentiment_category(polarity):
    if polarity > 0:
        return 'Positive'
    elif polarity < 0:
        
        return 'Negative'
    else:
        return 'Neutral'

# Polarity values matrum categories-ah dataframe-la map pannuvom
df['Polarity'] = df['Social_Media_Post'].apply(get_sentiment_polarity)
df['Sentiment'] = df['Polarity'].apply(get_sentiment_category)

print("\n--- Sentiment Results Summary ---")
print(df[['Social_Media_Post', 'Sentiment']])

# Step 4: Visualizing the Results (Warning illama clean-aa varradhuku)
plt.figure(figsize=(6, 4))
sns.countplot(x='Sentiment', data=df, hue='Sentiment', palette='Set2', legend=False)
plt.title('Social Media Data - Sentiment Distribution')
plt.xlabel('Sentiment Category')
plt.ylabel('Count')
plt.show()
```
